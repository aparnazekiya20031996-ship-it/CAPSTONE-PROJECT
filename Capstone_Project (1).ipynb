# **DATA PRE-PROCESSING**

# **Checking Missing Values**

### **Purpose**

### Identify the number of missing values in each column before performing data cleaning.


```python
print(df.isnull().sum())
```

### **Observation**

### Missing values are present in five columns. The Rating column has the highest number of missing values (1,474), while Type, Content Rating, Current Ver, and Android Ver contain only a small number of missing records.

# **Handling Missing Values**

### **Purpose**

### Handle the missing values identified during the data quality assessment using appropriate imputation methods.

```python
# Fill missing Rating values using the median
df["Rating"] = df["Rating"].fillna(df["Rating"].median())

# Fill missing categorical values using the mode
df["Type"] = df["Type"].fillna(df["Type"].mode()[0])
df["Content Rating"] = df["Content Rating"].fillna(df["Content Rating"].mode()[0])
df["Current Ver"] = df["Current Ver"].fillna(df["Current Ver"].mode()[0])
df["Android Ver"] = df["Android Ver"].fillna(df["Android Ver"].mode()[0])
```

```python
print(df.isnull().sum())
```

### **Observation**

### All missing values have been successfully handled. The dataset now contains **0 missing values across all 13 columns**.

# **Handling Duplicate Records**

### **Purpose**

###  Identify and remove duplicate records from the dataset to avoid repeated observations and improve the reliability of further analysis.

```python
# Identify duplicate rows
print("Number of duplicate rows:", df.duplicated().sum())

# Display duplicate rows
print("\nDuplicate rows:")
print(df[df.duplicated()])

# Remove duplicate rows
df = df.drop_duplicates()

# Display new shape
print("\nShape after removing duplicates:")
print(df.shape)

# Verify duplicates are removed
print("\nDuplicate rows remaining:")
print(df.duplicated().sum())
```

### **Observation**

### The dataset initially contained **483 duplicate records**. These duplicate rows were removed successfully, reducing the dataset from **10,841 records to 10,358 records**.

### After removal, the dataset contains **0 duplicate records**, confirming that the duplicate-handling step was completed successfully.

# **Identify Invalid Values**

### **Purpose**

### Identify invalid values in the Rating column by checking whether any ratings fall outside the valid range of 1 to 5. This helps ensure that incorrect values are removed before further analysis.


```python
# Identify invalid Rating values
print("Ratings greater than 5:", (df["Rating"] > 5).sum())
print("Ratings less than 1:", (df["Rating"] < 1).sum())
```

# **Handling Invalid Rating Values**

### **Purpose**

### Remove the invalid rating value that falls outside the valid 1–5 rating range and verify that only valid ratings remain in the dataset.

```python
# Remove invalid Rating values
df = df[(df["Rating"] >= 1) & (df["Rating"] <= 5)]

# Check the new shape
print("Shape after removing invalid ratings:", df.shape)

# Verify invalid ratings are removed
print("Invalid ratings remaining:", ((df["Rating"] < 1) | (df["Rating"] > 5)).sum())
```

### **Observation**

### One invalid rating value greater than 5 was identified and removed from the dataset. The dataset now contains **10,357 records**, and the validation check confirms that **no invalid rating values remain**.

# **Data Type Conversion**

### **Purpose**

### Convert the Reviews column into a numeric data type so that it can be used correctly for calculations, statistical analysis, and visualizations.

```python
# Convert Reviews to numeric
df["Reviews"] = pd.to_numeric(df["Reviews"], errors="coerce")

# Check the data type
print("Reviews data type:", df["Reviews"].dtype)
```

### **Observation**

### The Reviews column has been successfully converted from text format to the int64 numeric format. It can now be used for numerical calculations and analysis.

### **Convert Installs**

### **Purpose**

### Clean the Installs column by removing + and commas, then convert it from text to a numeric data type.

```python
# Clean and convert Installs to numeric
df["Installs"] = df["Installs"].str.replace("+", "", regex=False)
df["Installs"] = df["Installs"].str.replace(",", "", regex=False)
df["Installs"] = pd.to_numeric(df["Installs"], errors="coerce")

# Check the data type
print("Installs data type:", df["Installs"].dtype)
```

### **Observation**

### The Installs column has been successfully cleaned and converted from text format to the int64 numeric format. It can now be used for numerical calculations and analysis.

### **Convert Price**

### **Purpose**

### Remove the $ symbol from the Price column and convert it from text to a numeric data type.

```python
# Clean and convert Price to numeric
df["Price"] = df["Price"].str.replace("$", "", regex=False)
df["Price"] = pd.to_numeric(df["Price"], errors="coerce")

# Check the data type
print("Price data type:", df["Price"].dtype)
```

### **Observation**

### The Price column has been successfully cleaned and converted from text format to the float64 numeric format. It can now be used for numerical calculations and analysis.

### **Convert Size**

### **Purpose**

### Remove the M unit from the Size column and convert it from text to a numeric format for analysis.

```python
# Clean and convert Size to numeric
df["Size"] = df["Size"].str.replace("M", "", regex=False)
df["Size"] = pd.to_numeric(df["Size"], errors="coerce")

# Check the data type
print("Size data type:", df["Size"].dtype)
```

### **Observation**

### The Size column has been successfully cleaned and converted from text format to the float64 numeric format. It can now be used for numerical calculations and analysis.

### **Convert Last Updated**

### **Purpose**

### Convert the Last Updated column from text format to datetime format so that it can be used for date-based analysis, such as analyzing app updates by year.

```python
# Convert Last Updated to datetime
df["Last Updated"] = pd.to_datetime(df["Last Updated"])

# Check the data type
print("Last Updated data type:", df["Last Updated"].dtype)
```

### **Observation**

### The Last Updated column has been successfully converted from text format to the datetime64[ns] format. It can now be used for date-based analysis, such as identifying app update trends over time.


### **Handling Variable App Size**

### **Purpose**

### Create an indicator column to identify apps whose original size was recorded as "Varies with device".

```python
# Create an indicator for apps with variable size
df["Size_Varies"] = df["Size"].isna().astype(int)

# Check the number of apps with variable size
print("Apps with variable size:", df["Size_Varies"].sum())
```

### **Observation**

### A total of 1,841 apps have sizes that vary by device. The Size_Varies indicator was created to preserve this information instead of treating these values as ordinary missing data.

# **Data Validation**

### **Purpose**

### Verify that the main cleaning steps were completed successfully and check the current structure of the cleaned dataset before moving to further analysis.

```python
# Check the cleaned dataset

print("Dataset Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)
```

### **Observation**

### The data validation confirms that the cleaned dataset contains **10,357 records and 14 columns**, with **no duplicate records** and no missing values except for the Size column.

### The Size missing values are expected because they represent apps with **"Varies with device"** sizes, which are captured separately using the Size_Varies indicator. The columns have also been converted to appropriate data types, confirming that the dataset is ready for further analysis.

# **Creating New Features**

## **Reviews per Install**

### **Purpose**

### Create a new Reviews_per_Install column to measure the proportion of installations that resulted in user reviews.

```python
# Calculate reviews per install
df["Reviews_per_Install"] = df["Reviews"] / df["Installs"]

# Display the new feature
print(df["Reviews_per_Install"].head())
```

### **Observation**

### The Reviews_per_Install feature was successfully created by dividing the number of reviews by the number of installs. This feature provides a simple measure of user engagement relative to the number of installations.

## **Estimated Revenue**

### **Purpose**

### Create an Estimated_Revenue feature by multiplying the number of installs by the listed app price.

```python
# Calculate estimated revenue
df["Estimated_Revenue"] = df["Installs"] * df["Price"]

# Display the new feature
print(df["Estimated_Revenue"].head())
```

Observation

### The Estimated_Revenue feature was successfully created by multiplying Installs by Price. Free apps have an estimated revenue of **0**, while paid apps can have a positive estimated value.



## **Updated Year**

### **Purpose**

### Extract the year from the Last Updated column to make it easier to analyze app update trends over time.

```python
# Extract the year from Last Updated
df["Updated Year"] = df["Last Updated"].dt.year

# Display the new feature
print(df["Updated Year"].head())
```

### **Observation**

### The Updated Year feature was successfully created by extracting the year from the Last Updated column. This feature will be useful for analyzing app update trends over time.

## **Feature Validation**

### **Purpose**

### Verify that the newly created features were added correctly and check for any missing values generated during feature creation.

```python
# Display the newly created features
print(df[[
    "Size_Varies",
    "Reviews_per_Install",
    "Estimated_Revenue",
    "Updated Year"
]].head())

# Check missing values in the new features
print("\nMissing values in new features:")
print(df[[
    "Size_Varies",
    "Reviews_per_Install",
    "Estimated_Revenue",
    "Updated Year"
]].isnull().sum())
```

### **Observation**

### The newly created features were successfully added to the dataset.

### - Size_Varies, Estimated_Revenue, and Updated Year contain no missing values.
### - Reviews_per_Install contains **15 missing values** because some apps have zero installations, making the calculation Reviews / Installs undefined.
### - These 15 values will be retained as missing rather than artificially imputed.
